### 준비: 데이터 불러오기
이 노트에서 `df`를 쓰려면 아래 셀을 먼저 실행해요. (실습 1-3과 같은 코드)

````python
import pandas as pd   # pandas를 불러오고 짧게 pd라고 부르기로 함
url = "https://loreenkim0301.github.io/swy-lecturenote/261026-ai-pm-week1/data/04_orders.csv"   # 데이터 주소
df = pd.read_csv(url)   # CSV를 읽어 표(df)로 저장
df.head()               # 앞 5줄만 미리 보기
````

### 실습
아래 코드 셀을 실행해요.

````python
df.groupby("채널")["매출"].sum()   # 채널별로 묶어서 매출을 모두 더하기
````

**실행 결과 · 이렇게 나오면 맞아요**


### 실습
아래 코드 셀을 실행해요.

````python
df.groupby("상품")["수량"].sum().sort_values(ascending=False)   # 상품별 수량 합계를 큰 순서로 정렬
````
