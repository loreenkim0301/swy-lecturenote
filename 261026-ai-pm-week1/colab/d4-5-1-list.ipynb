### 🎮게임 개발자는 리스트를 이렇게 써요
가방(인벤토리)의 아이템을 리스트 하나에 순서대로 담아요.

### 실습
아래 코드를 새 코드 셀에 붙여 넣고 실행해요.

```python
inventory = ["포션", "나무 검", "방패"]   # 가방(인벤토리)
inventory.append("열쇠")                 # 가방 끝에 열쇠 넣기
print("첫 번째 칸:", inventory[0])       # 0번째 = 첫 번째 칸
print("아이템 개수:", len(inventory), inventory)
```

### 더 해 보기: 아이템 팔기 새 코드 셀에서 실행해요. 맨 뒤 칸을 꺼내 보고, 리스트에서 값을 빼요.

```python
inventory = ["포션", "나무 검", "방패", "열쇠"]   # 가방
print("마지막 칸:", inventory[-1])     # -1은 맨 뒤 칸
inventory.remove("나무 검")            # 나무 검을 팔아 가방에서 뺌
print(inventory)
```
