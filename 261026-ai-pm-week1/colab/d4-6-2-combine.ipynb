### 실습
지금까지의 다섯 가지를 한 번에 쓰면 전투 화면의 로직이 돼요. 몬스터 목록을 보여 주고, 이번에 공격할 몬스터를 무작위로 골라요.

```python
import random                    # 무작위 기능 불러오기
monsters = [                     # 몬스터 정보 묶음 여러 개
    {"name": "슬라임", "hp": 30},
    {"name": "고블린", "hp": 0},
    {"name": "오크", "hp": 50},
]
for m in monsters:               # 몬스터를 하나씩 꺼내 반복
    if m["hp"] == 0:             # 체력이 0이면
        print(m["name"], "- 쓰러짐")
    else:                        # 아니면 남은 체력 표시
        print(m["name"], "- 체력", m["hp"])
target = random.choice(monsters) # 공격 대상을 무작위로 고름
print("이번 공격 대상:", target["name"])
```
