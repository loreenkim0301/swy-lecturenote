### 🎮게임 개발자는 변수를 이렇게 써요
캐릭터의 이름·체력·돈을 변수에 저장해 두고, 싸울 때마다 값을 바꿔요. 오늘 실습은 모두 게임 개발자가 실제로 쓰는 모양의 코드로 해요.

### 실습
아래 코드를 새 코드 셀에 붙여 넣고 실행해요.

```python
player_name = "용사"     # 캐릭터 이름 (글자는 따옴표로 감싸요)
hp = 100                 # 체력 (숫자는 따옴표 없이)
gold = 500               # 가진 돈
hp = hp - 30             # 몬스터에게 30만큼 피해를 입음
print(player_name, "남은 체력:", hp, "/ 골드:", gold)
```

### 실습
같은 셀에서 hp = hp - 30의 30을 지우고 50으로 바꿔 다시 실행해 보세요.

### 더 해 보기: 상점에서 포션 사기 새 코드 셀에서 실행해요. 곱하기(*)로 포션값을 계산하고, 가진 돈에서 빼요.

```python
gold = 500                    # 가진 돈
potion_price = 50             # 포션 한 개 가격
qty = 3                       # 살 개수
cost = potion_price * qty     # 포션값 = 가격 × 개수
gold = gold - cost            # 가진 돈에서 포션값을 뺌
print("포션", qty, "개 구매 / 쓴 돈:", cost, "/ 남은 골드:", gold)
```
