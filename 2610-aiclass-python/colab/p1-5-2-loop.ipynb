### 🎮게임 개발자는 반복을 이렇게 써요
몬스터 여러 마리에게 같은 공격을 하나씩 반복해요. 몬스터가 3마리든 300마리든 코드는 그대로예요.

### 실습
코드는 한 번만 썼는데 결과가 몇 줄 나오는지 보세요.

```python
monsters = ["슬라임", "고블린", "오크"]   # 이번 웨이브의 몬스터
for m in monsters:                       # 한 마리씩 꺼내 m에 담고 반복
    print("용사가", m, "을(를) 공격했어요!")
print("웨이브 클리어!")                   # 반복이 끝난 뒤 한 번만
```

### 실습
같은 셀에서 리스트 끝에 "드래곤"을 추가해 ["슬라임", "고블린", "오크", "드래곤"]으로 바꾸고 다시 실행해 보세요. 공격 줄이 한 줄 늘어나면 맞아요.

### 더 해 보기: 웨이브 경험치 합계 구하기 반복하면서 값을 하나씩 더해 가는 방법이에요.

```python
exps = [10, 25, 40]          # 몬스터별 경험치
total = 0                    # 합계를 0으로 시작
for e in exps:               # 경험치를 하나씩 꺼내
    total = total + e        # 지금까지의 합계에 더함
print("웨이브 경험치 합계:", total)
print(sum(exps))             # 합계 기능으로 다시 계산
```
