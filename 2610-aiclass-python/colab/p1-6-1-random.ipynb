### 🎮게임 개발자는 random을 이렇게 써요
공격 피해량과 아이템 드롭처럼 매번 결과가 달라야 하는 곳에 random을 써요.

### 실습
아래 코드 셀을 여러 번 실행해 보세요.

```python
import random                               # 무작위 기능 묶음 불러오기
damage = random.randint(10, 20)             # 10~20 사이 피해
drop = random.choice(["포션", "금화", "꽝"])  # 떨어뜨리는 아이템
print("피해:", damage, "/ 드롭 아이템:", drop)
```

### 더 해 보기: 치명타 확률 만들기 random과 조건을 함께 써요. 1~100 중 20 이하가 나오면 치명타(확률 20%)예요.

```python
import random                     # 무작위 기능 불러오기
roll = random.randint(1, 100)     # 1~100 중 숫자 하나 뽑기
if roll <= 20:                    # 20 이하면 (확률 20%)
    print(roll, "→ 치명타! 피해 2배")
else:
    print(roll, "→ 일반 공격")
```
