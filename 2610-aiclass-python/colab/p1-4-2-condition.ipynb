### 🎮게임 개발자는 조건을 이렇게 써요
체력에 따라 게임 오버, 경고, 계속 진행 중 하나를 고르는 규칙을 if로 만들어요.

### 실습
아래 코드 셀을 실행해요.

```python
hp = 15                  # 지금 체력
if hp <= 0:              # 체력이 0 이하이면
    print("게임 오버")
elif hp < 20:            # 아니고, 20보다 적으면
    print("체력이 부족해요! 포션을 마시세요")
else:                    # 둘 다 아니면
    print("모험을 계속해요")
```

### 실습
같은 셀에서 hp = 15를 0, 50으로 바꿔 가며 결과를 비교해 보세요.

### 더 해 보기: 보스 방 입장 조건 만들기 게임 규칙은 "레벨 10 이상이면 보스 방에 들어갈 수 있다"예요. "이상"은 >=로 써요.

```python
level = 9                    # 지금 레벨
if level >= 10:              # 10 "이상"이면 (>= 크거나 같다)
    print("보스 방에 들어갈 수 있어요")
else:                        # 그렇지 않으면
    print("레벨", 10 - level, "만 더 올리면 들어갈 수 있어요")
```
